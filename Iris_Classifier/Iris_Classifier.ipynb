```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Iris.csv")
```

```python
df = df.rename(columns={"SepalLengthCm" : "Sepal_length" , "PetalWidthCm" : "Petal_width" , "SepalWidthCm" : "Sepal_width" ,"PetalLengthCm" : "Petal_length"})
```

```python
df['Species'].value_counts()
```

```python
df.drop(columns=["Id"], inplace = True)
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix
```

```python
X = df.drop(columns=["Species"])
y = df["Species"]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = None, test_size=0.2)
```

```python
print(X_train.shape , X_test.shape)
```

```python
print(y_train.shape , y_test.shape)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
X_train_pred = model.predict(X_train)
training_accuracy = accuracy_score(X_train_pred, y_train)
```

```python
print("Accuracy on Training data : ", training_accuracy*100)
```

```python
X_test_pred = model.predict(X_test)
test_accuracy = accuracy_score(X_test_pred, y_test)
```

```python
print(classification_report(y_test, X_test_pred))
```

```python
print(confusion_matrix(y_test, X_test_pred))
```

```python
print("Accuracy on Testing data : ", test_accuracy*100)
```

```python
input_dict = {
    'Sepal_length': 6.8,
    'Sepal_width':  3.5,
    'Petal_length': 5.2,
    'Petal_width': 1.4
}
```
