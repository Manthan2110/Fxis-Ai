List, Tuple, Set, Dictionary 

```python
# list should be included in the square brackets
list = [1,2,3,4,5]
print(list)
type(list)
```

```python
# list have multiple data types
my_list = [2,3,1.8, 'Manthan', True]
print(my_list)
```

Tuple

```python
# tuple should be included in the round bracket
tuple = (1,2,3,4)
print(tuple)
```

```python
# tuple allows multiple datatypes
tuple = (1,2,3, "Manthan", True)
print(tuple)
```

```python
# tuple allows duplicates values
tuple= (1,1,2,3,4,5,4)
print(tuple)
```

```python
#coverting a list into a tuple
my_list = [2,3,1.8, 'Manthan', True, 2,3]
print(my_list)

my_tuple = tuple(my_list)
print(my_tuple)
```

```python
tuple = (1,2,3, "Manthan", True)
print(tuple[0])
print(tuple[3])
```

Tuples are immutable --> Unchangeable 

```python
print(len(tuple))
```

Set

```python
#set should be in Curly Brackets.
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
# covert list into set
list1 = [1,2,3,4,5]

x = set(list1)
print(x)
```

```python
# set does not allow duplicate values
my_set = {1,2,3,4,5,1,2,3}
print(my_set)
```

Dictionary

Key-Value Pair

```python
my_dictionary = {'name':'David','age':30,'country':'India'}
print(my_dictionary)
type(my_dictionary)
```

```python
print(my_dictionary['name'])
print(my_dictionary['age'])
print(my_dictionary['country'])
```

```python
# dictionary does not allow duplicate values
dictionary_2 = {'name':'David','age':30,'country':'India','name':'David','age':30,'country':'India'}
print(dictionary_2)
```
